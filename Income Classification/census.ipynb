# Task
The task is to apply dimensionality reduction using Principal Component Analysis (PCA) on the preprocessed `X` DataFrame. This involves identifying and separating categorical and numerical features, applying one-hot encoding to categorical features, scaling numerical features using `StandardScaler`, combining the processed features, and then applying PCA to reduce dimensionality while retaining most of the variance. ## Identify Feature Types

### Subtask:
Separate the features in the DataFrame 'X' into categorical and numerical types.


**Reasoning**:
The first step is to inspect the data types of columns in DataFrame `X` to differentiate between categorical and numerical features. `X.dtypes` is suitable for this purpose.



````python
X.dtypes
````

**Reasoning**:
Based on the output of X.dtypes, I will now separate the column names into two lists: one for categorical features (type 'object') and one for numerical features (type 'int64').



````python
categorical_features = X.select_dtypes(include=['object']).columns.tolist()
numerical_features = X.select_dtypes(include=['int64', 'float64']).columns.tolist()

print("Categorical Features:", categorical_features)
print("Numerical Features:", numerical_features)
````

## Encode Categorical Features

### Subtask:
Apply one-hot encoding to the categorical features in 'X' to convert them into a numerical format.


**Reasoning**:
First, I'll import the `OneHotEncoder` from `sklearn.preprocessing` to prepare for encoding the categorical features.



````python
from sklearn.preprocessing import OneHotEncoder
````

**Reasoning**:
Now I will instantiate `OneHotEncoder` with `handle_unknown='ignore'` and then fit and transform the `categorical_features` from `X`.



````python
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
X_encoded = encoder.fit_transform(X[categorical_features])
````

**Reasoning**:
Now I will create a new DataFrame from the one-hot encoded features using the feature names obtained from the encoder.



````python
import pandas as pd
X_encoded_df = pd.DataFrame(X_encoded, columns=encoder.get_feature_names_out(categorical_features))
X_encoded_df.head()
````

## Scale Numerical Features

### Subtask:
Scale the numerical features in 'X' using `StandardScaler` to normalize their ranges.

**Reasoning**:
To prepare for scaling, I will import `StandardScaler` from `sklearn.preprocessing`.



````python
from sklearn.preprocessing import StandardScaler
````

**Reasoning**:
Now I will instantiate `StandardScaler` and then fit and transform the numerical features from `X`.



````python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X[numerical_features])
````

**Reasoning**:
Now I will create a new DataFrame from the scaled numerical features for easier combination with the encoded categorical features.



````python
X_scaled_df = pd.DataFrame(X_scaled, columns=numerical_features)
X_scaled_df.head()
````

**Reasoning**:
I need to combine the one-hot encoded categorical features (`X_encoded_df`) and the scaled numerical features (`X_scaled_df`) into a single DataFrame to prepare for PCA.



````python
X_processed = pd.concat([X_encoded_df, X_scaled_df], axis=1)
X_processed.head()
````

## Apply PCA for Dimensionality Reduction

### Subtask:
Apply PCA to the `X_processed` DataFrame to reduce its dimensionality while retaining a significant portion of the variance. Determine the number of components needed to explain at least 95% of the variance.

**Reasoning**:
To perform PCA, I first need to import the `PCA` class from `sklearn.decomposition`.



````python
from sklearn.decomposition import PCA
````

**Reasoning**:
Now, I will instantiate `PCA` with `n_components=0.95` to retain 95% of the variance, fit it to the `X_processed` DataFrame, and then transform the data.



````python
pca = PCA(n_components=0.95)
X_pca = pca.fit_transform(X_processed)
````

**Reasoning**:
After fitting PCA, I will print the number of components selected and their explained variance ratio to summarize the dimensionality reduction.



````python
print(f"Number of components selected: {pca.n_components_}")
print(f"Total explained variance ratio: {pca.explained_variance_ratio_.sum():.2f}")
````

Split the `X_pca` and `y` data into training and testing sets to prepare for model training and evaluation.

## Split Data

### Subtask:
Split the `X_pca` (dimensionally reduced features) and `y` (target variable) into training and testing sets to prepare for model training and evaluation.


**Reasoning**:
To split the data, I need to import the `train_test_split` function from `sklearn.model_selection`.



````python
from sklearn.model_selection import train_test_split
````

**Reasoning**:
Now I will split `X_pca` and `y` into training and testing sets using the specified parameters: `test_size=0.2`, `random_state=42`, and `stratify=y`.



````python
X_train, X_test, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=42, stratify=y)
````

## Train Classification Model

### Subtask:
Choose and train a suitable classification model (e.g., Logistic Regression or RandomForestClassifier) on the training data.


**Reasoning**:
To train a classification model, I need to first import the chosen model class from `sklearn`. I will use `LogisticRegression` for this purpose.



````python
from sklearn.linear_model import LogisticRegression
````

**Reasoning**:
Now I will instantiate `LogisticRegression` with `random_state=42` for reproducibility and then train the model using `X_train` and `y_train`.



````python
model = LogisticRegression(random_state=42, max_iter=1000)
model.fit(X_train, y_train)
````

**Reasoning**:
The previous step produced a DataConversionWarning because `y_train` is a column-vector, but `LogisticRegression` expects a 1D array. I will fix this by applying `.ravel()` to `y_train` when fitting the model.



````python
model = LogisticRegression(random_state=42, max_iter=1000)
model.fit(X_train, y_train.values.ravel())
````

## Evaluate Model Performance

### Subtask:
Evaluate the trained classification model using the test data and calculate appropriate metrics such as accuracy, precision, recall, and F1-score.


**Reasoning**:
To evaluate the model's performance, the first step is to use the trained model to make predictions on the test dataset.



````python
y_pred = model.predict(X_test)
````

**Reasoning**:
Now, I will import the necessary evaluation metrics (`accuracy_score`, `precision_score`, `recall_score`, and `f1_score`) from `sklearn.metrics` to assess the model's performance.



````python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
````

**Reasoning**:

````python
y_pred = model.predict(X_test)
````

**Reasoning**:
Now that the data has been re-split and predictions have been updated, I will calculate and print the accuracy, precision, recall, and F1-score of the model using the correctly formatted `y_test` and the new `y_pred`.

## Create Streamlit Application File

### Subtask:
Create a new Python file (e.g., app.py) for the Streamlit application. This file will contain all the necessary code to load the model and preprocessors, accept user input, perform preprocessing, make predictions, and display results.


**Reasoning**:
The first step in creating the Streamlit application is to create the `app.py` file and include the necessary imports and component loading. I will use the `%%writefile` magic command to create this file and populate it with initial code that loads the previously saved models and preprocessors.



````python
!pip install streamlit
````

````python
%%writefile app.py
import streamlit as st
import joblib
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression

# Load the exported model and preprocessing components
model = joblib.load('logistic_regression_model.joblib')
encoder = joblib.load('one_hot_encoder.joblib')
scaler = joblib.load('standard_scaler.joblib')
pca = joblib.load('pca_model.joblib')

# Define feature lists (these were identified during preprocessing)
categorical_features = ['workclass', 'education', 'marital-status', 'occupation', 'relationship', 'race', 'sex', 'native-country']
numerical_features = ['age', 'fnlwgt', 'education-num', 'capital-gain', 'capital-loss', 'hours-per-week']

st.title('Adult Income Prediction')
````

**Reasoning**:
Now I will add the `preprocess_input` function to `app.py`. This function will take user input, convert it into a DataFrame, apply one-hot encoding to categorical features, scale numerical features, combine them, and then apply PCA using the loaded components. This function is critical for transforming raw user input into a format suitable for the trained model.



````python
%%writefile -a app.py
def preprocess_input(input_df):
    # One-hot encode categorical features
    input_encoded = encoder.transform(input_df[categorical_features])
    input_encoded_df = pd.DataFrame(input_encoded, columns=encoder.get_feature_names_out(categorical_features))

    # Scale numerical features
    input_scaled = scaler.transform(input_df[numerical_features])
    input_scaled_df = pd.DataFrame(input_scaled, columns=numerical_features)

    # Concatenate processed features
    input_processed = pd.concat([input_encoded_df, input_scaled_df], axis=1)

    # Apply PCA
    input_pca = pca.transform(input_processed)
    return input_pca
````

**Reasoning**:
Now I will add the Streamlit UI elements to `app.py` for user input. This involves creating input widgets for all categorical and numerical features, populating `st.selectbox` with unique options for categorical features, and adding a prediction button. I will hardcode the unique values for categorical features to ensure the app functions correctly, as these were implicitly learned during the `OneHotEncoder` fit.



````python
%%writefile -a app.py
# Streamlit UI for user input
st.sidebar.header('User Input Features')

def user_input_features():
    # Get unique values for categorical features from the encoder (assuming they were fitted properly)
    # This is a workaround as encoder.categories_ returns arrays, which need to be converted to lists/tuples

    # For 'workclass'
    workclass_options = encoder.categories_[categorical_features.index('workclass')].tolist()
    workclass = st.sidebar.selectbox('Workclass', workclass_options)

    # For 'education'
    education_options = encoder.categories_[categorical_features.index('education')].tolist()
    education = st.sidebar.selectbox('Education', education_options)

    # For 'marital-status'
    marital_status_options = encoder.categories_[categorical_features.index('marital-status')].tolist()
    marital_status = st.sidebar.selectbox('Marital Status', marital_status_options)

    # For 'occupation'
    occupation_options = encoder.categories_[categorical_features.index('occupation')].tolist()
    occupation = st.sidebar.selectbox('Occupation', occupation_options)

    # For 'relationship'
    relationship_options = encoder.categories_[categorical_features.index('relationship')].tolist()
    relationship = st.sidebar.selectbox('Relationship', relationship_options)

    # For 'race'
    race_options = encoder.categories_[categorical_features.index('race')].tolist()
    race = st.sidebar.selectbox('Race', race_options)

    # For 'sex'
    sex_options = encoder.categories_[categorical_features.index('sex')].tolist()
    sex = st.sidebar.selectbox('Sex', sex_options)

    # For 'native-country'
    native_country_options = encoder.categories_[categorical_features.index('native-country')].tolist()
    native_country = st.sidebar.selectbox('Native Country', native_country_options)

    age = st.sidebar.slider('Age', 17, 90, 30)
    fnlwgt = st.sidebar.slider('Final Weight (fnlwgt)', 10000, 1500000, 200000)
    education_num = st.sidebar.slider('Education Number', 1, 16, 9)
    capital_gain = st.sidebar.slider('Capital Gain', 0, 100000, 0)
    capital_loss = st.sidebar.slider('Capital Loss', 0, 4356, 0)
    hours_per_week = st.sidebar.slider('Hours per Week', 1, 99, 40)

    data = {
        'age': age,
        'workclass': workclass,
        'fnlwgt': fnlwgt,
        'education': education,
        'education-num': education_num,
        'marital-status': marital_status,
        'occupation': occupation,
        'relationship': relationship,
        'race': race,
        'sex': sex,
        'capital-gain': capital_gain,
        'capital-loss': capital_loss,
        'hours-per-week': hours_per_week,
        'native-country': native_country
    }
    features = pd.DataFrame(data, index=[0])
    return features

input_df = user_input_features()

st.subheader('User Input features')
st.write(input_df)

if st.sidebar.button('Predict Income'):
    processed_input = preprocess_input(input_df)
    prediction = model.predict(processed_input)

    st.subheader('Prediction')
    st.write('The predicted income is: ')
    if prediction[0] == '>50K':
        st.write('**>50K**')
    else:
        st.write('**<=50K**')
````
